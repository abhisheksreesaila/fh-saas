# Tests — utils_ai_interface

```
from fh_saas.utils_ai_interface import (
    AIInterfaceConfig,
    validate_ai_interface_config,
    Interface,
)


def _dummy_fn(text: str):
    return {'echo': text}
```

```
cfg = AIInterfaceConfig(max_concurrency=1, default_poll_interval_ms=1200)
errs = validate_ai_interface_config(cfg)
assert errs == [], f'Expected no errors, got: {errs}'
print('✅ config validation passes')
```

```
bad = AIInterfaceConfig(max_concurrency=0)
errs = validate_ai_interface_config(bad)
assert any('max_concurrency' in e for e in errs)
print('✅ invalid config catches max_concurrency')
```

```
iface = Interface(fn=_dummy_fn, inputs=['text'], outputs=['text'])
meta = iface.describe()
assert meta['input_count'] == 1
assert meta['output_count'] == 1
assert meta['max_concurrency'] >= 1
print('✅ interface describe works')
```

```
contract = iface.build_polling_contract('job_123')
assert contract['status_endpoint'].endswith('/api/jobs/job_123')
assert contract['result_endpoint'].endswith('/api/jobs/job_123/result')
assert contract['poll_interval_ms'] >= 100
print('✅ polling contract builds correctly')
```
